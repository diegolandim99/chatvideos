## 🚀 Como Usar
1. Clique em **Ambiente de execução → Executar tudo** (ou `Ctrl+F9`)
2. Aguarde ~3 min enquanto as dependências são instaladas
3. Clique em **🎬 ABRIR O CLIPAI** ao final

> **Dica:** Conecte ao Google Drive quando solicitado para salvar seus clips automaticamente.

```python
# @title 📁 Passo 1 — Conectar ao Google Drive (opcional, mas recomendado)
try:
    from google.colab import drive
    drive.mount('/content/drive')
    print('✅ Google Drive conectado. Clips salvos em: Meu Drive/ClipAI/')
except Exception:
    print('⚠️ Rodando fora do Colab. Clips salvos em ~/ClipAI/')
```

```python
# @title ⚙️ Passo 2 — Instalar dependências do sistema
import subprocess, sys
from IPython.display import display, HTML, clear_output

display(HTML('<p style="color:#4fc3f7;font-family:monospace;font-size:14px">⏳ ClipAI — instalando dependências...</p>'))

cmds = [
    ['apt-get', 'install', '-y', '-q', 'ffmpeg'],
    ['apt-get', 'install', '-y', '-q', 'ttyd'],
    [sys.executable, '-m', 'pip', 'install', '-q', 'yt-dlp'],
    [sys.executable, '-m', 'pip', 'install', '-q', 'openai-whisper'],
    [sys.executable, '-m', 'pip', 'install', '-q', 'srt'],
]

nomes = ['ffmpeg', 'ttyd', 'yt-dlp', 'openai-whisper', 'srt']

for nome, c in zip(nomes, cmds):
    r = subprocess.run(c, capture_output=True, text=True)
    status = '✅' if r.returncode == 0 else '❌'
    print(f'{status}  {nome}')

clear_output(wait=True)
print('✅ Dependências instaladas com sucesso!')
```

```python
# @title 📦 Passo 3 — Baixar o ClipAI
import os, sys, subprocess

REPO_URL = "https://github.com/diegolandim99/chatvideos.git"
WORK_DIR = "/tmp/clipai"

if os.path.exists(WORK_DIR):
    subprocess.run(
        ["git", "-C", WORK_DIR, "pull", "--ff-only", "--depth", "1"],
        capture_output=True
    )
    print("🔄 ClipAI atualizado.")
else:
    subprocess.run(
        ["git", "clone", "--depth", "1", REPO_URL, WORK_DIR],
        capture_output=True
    )
    print("✅ ClipAI baixado.")

os.chdir(WORK_DIR)
sys.path.insert(0, WORK_DIR)
```

```python
# @title 🎬 Passo 4 — Iniciar o ClipAI
import subprocess, threading, time
from IPython.display import display, HTML

PORT = 7681

def start_terminal():
    subprocess.Popen(
        [
            "ttyd", "-p", str(PORT), "-W",
            "bash", "-i", "-c",
            f"cd /tmp/clipai && python main.py; exec bash"
        ],
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    )

t = threading.Thread(target=start_terminal, daemon=True)
t.start()
time.sleep(2)

from google.colab.output import eval_js
url = eval_js(f"google.colab.kernel.proxyPort({PORT})")

html = f"""<div style="
  background: linear-gradient(135deg, #0f0c29, #302b63, #24243e);
  border-radius: 12px;
  padding: 28px 32px;
  text-align: center;
  font-family: 'Google Sans', Arial, sans-serif;
  margin-top: 12px;
">
  <div style="font-size: 42px; margin-bottom: 8px;">🎬</div>
  <div style="color: #fff; font-size: 22px; font-weight: 700; margin-bottom: 4px;">
    ClipAI está pronto!
  </div>
  <div style="color: #aaa; font-size: 13px; margin-bottom: 20px;">
    YouTube Shorts · TikTok · Instagram Reels
  </div>
  <a href="{url}" target="_blank" style="
    display: inline-block;
    background: linear-gradient(90deg, #f953c6, #b91d73);
    color: white;
    text-decoration: none;
    padding: 14px 36px;
    border-radius: 30px;
    font-size: 16px;
    font-weight: 700;
    letter-spacing: 0.5px;
    box-shadow: 0 4px 20px rgba(249,83,198,0.4);
  ">🚀 ABRIR O CLIPAI</a>
  <div style="color: #666; font-size: 11px; margin-top: 16px;">
    Todos os clips são salvos em:
    <code style="color:#f953c6">Meu Drive/ClipAI/</code>
  </div>
</div>"""

display(HTML(html))
```
